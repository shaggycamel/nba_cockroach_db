# Attempt to obtain FTY box scores

```python
for period in list(range(1, 30)):
    box_score = fty_con.box_scores(matchup_period = period)
    for mtch_up in box_score:
        for h_a in ['home', 'away']:
            
            competitor = getattr(mtch_up, h_a + '_team')
            print('period:', period, '/ team_id:', competitor.team_id)

            competitor_stats = getattr(mtch_up, h_a + '_stats')
            for key in competitor_stats.keys():
                print(key, competitor_stats[key]['value'])
            print()
```
